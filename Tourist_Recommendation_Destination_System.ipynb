# 🏖️ Tourist Destination Recommendation System

# Introduction

Tourism is one of the fastest-growing industries in the world, providing people with opportunities to explore new places, experience different cultures, and create memorable travel experiences. With thousands of tourist destinations available across India, selecting the most suitable destination can be difficult because every traveler has different preferences, budgets, interests, and travel requirements.

A Tourist Destination Recommendation System helps users discover destinations that best match their preferences by analyzing various factors such as destination category, travel budget, ideal season, ratings, travel duration, and suitability for families, couples, solo travelers, or groups of friends. Instead of manually searching through multiple websites, users receive personalized recommendations based on their interests.

This project applies Data Science and Machine Learning techniques to analyze tourism data, clean and preprocess the dataset, perform exploratory data analysis, engineer meaningful features, and build a recommendation system. The project demonstrates the complete data science workflow from raw data to generating personalized travel recommendations.

# Objective

The primary objective of this project is to develop a Tourist Destination Recommendation System that suggests suitable tourist destinations based on user preferences and destination characteristics.

The project aims to:

## Fill the missing values in the Dataframe

### **Suitable_For**

A new feature, **`Suitable_For`**, was created by combining the `Family_Friendly`, `Couple_Friendly`, `Solo_Friendly`, `Friends_Friendly`, `Kids_Friendly`, and `Senior_Citizen_Friendly` columns into a single descriptive column. The new column identifies the type(s) of travelers for whom each destination is suitable, such as **Family**, **Couple**, **Solo Travelers**, **Friends**, **Kids**, and **Senior Citizens**. This feature provides a concise summary of destination suitability, making the dataset easier to understand while improving data analysis, visualization, and personalized travel recommendations.

```python
def suitable_for(row):
    categories=[]
    if row['Family_Friendly'] == 'yes':
        categories.append('Family')
    if row['Couple_Friendly'] =='yes':
        categories.append('Couple')
    if row['Solo_Friendly'] == 'yes':
        categories.append('Solo')
    if row['Friends_Friendly'] == 'yes':
        categories.append('Friends')
    if row['Kids_Friendly'] =='yes':
        categories.append('Kids')
    if row['Senior_Citizen_Friendly']=='yes':
        categories.append('Senior Citizen')
    return ", ".join(categories)
df['Suitable_For'] = df.apply(suitable_for,axis=1)
print(df[['Destination_Name','State','City','Suitable_For']].head(20))
```

**Distribution by Session**

```python
month_order = [
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
]

month_map = {
    'January':'Jan', 'February':'Feb', 'March':'Mar',
    'April':'Apr', 'May':'May', 'June':'Jun',
    'July':'Jul', 'August':'Aug', 'September':'Sep',
    'October':'Oct', 'November':'Nov', 'December':'Dec'
}

months = []

for value in df['Recommended_Months'].dropna():
    value = str(value)

    
    for full, short in month_map.items():
        value = value.replace(full, short)

    value = value.replace(" to ", "-")
    value = value.replace(" ", "")

    
    if "-" in value:
        start, end = value.split("-")
        start_index = month_order.index(start)
        end_index = month_order.index(end)
        months.extend(month_order[start_index:end_index+1])

    
    elif "," in value:
        months.extend(value.split(","))

    
    else:
        months.append(value)


month_counts = pd.Series(months).value_counts().reindex(month_order, fill_value=0)

plt.figure(figsize=(10,5))

sns.barplot(
    x=month_counts.index,
    y=month_counts.values,
    palette="viridis"
)

plt.title("Most Recommended Months for Tourist Destinations")
plt.xlabel("Month")
plt.ylabel("Number of Destinations")

plt.show()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
features_matrix = cv.fit_transform(df['Combined_Features'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(features_matrix)
```

```python
def recommend(destination_name):
    index = df[df['Destination_Name'] == destination_name].index[0]
    distances = list(enumerate(similarity[index]))
    sorted_destinations = sorted(
        distances,
        key=lambda x: x[1],
        reverse=True
    )[1:6]
    for i in sorted_destinations:
        print(df.iloc[i[0]]['Destination_Name'])
```
